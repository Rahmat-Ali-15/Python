## Lambda Function

```python
def add(a: int,b: int) -> int:
    return a + b

a = int(input())
b = int(input())

add(a,b)
```

```python
add = lambda a, b : a + b

add(2,7)
```

```python
add = lambda n : True if n % 2 == 0 else False

add(4)
```

```python
from typing import List

# Function Declaration & Defination
def firstIndex(li: List[int], n: int) -> int:
    for index in range(len(li)):
        if li[index] == n:
            return index
    else:
        return -1

# Talking input from the user
n = int(input("Enter the number : "))
li = [ int(item) for item in input("Enter the list element : ").split() ]

# Function call
index = firstIndex(li, n)

# Output Display

if index == -1 :
    print(f"{n} not found in {li}")
else:
    print(f"{n} found in {li} at index {index}")
```
