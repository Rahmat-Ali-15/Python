```python
def add2(a: int, b: int) -> int:
    return a + b

a = int(input("Entet the first number : "))
b = int(input("Enter the second number : "))

total = add2(a,b)

print(f"The sum of {a} & {b} is {total}")
```

```python
def divide(a: int, b : int) -> float | str:
    if b == 0:
        return "Denominator Cannot be Zero!"
    return a/b

divide(10,0)
# divide(10,2)
```

```python
n = int(input())

if n % 400 != 0 and (n % 4 == 0 and n % 100 == 0) :
    print(f"{n} Not a Leap year")
elif (n % 4 == 0 and n % 100 != 0) or n % 400 == 0:
    print(f"{n} Leap year")
else:
    print(f" {n} Not a Leap year")
```
