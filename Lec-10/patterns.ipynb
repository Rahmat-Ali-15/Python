# Patterns

```python
n = 5

for i in range(n):
    for j in range(n):
        print("*", end="")
    print()
```

```python
n = 4

for i in range(n):
    for j in range(i+1):
        print("*", end="")
    print()
```

```python
n = 5

for i in range(n):
    for j in range(n):
        print(i+1, end=" ")
    print()
```

```python
n = 5

for i in range(n):
    for j in range(n):
        print(j+1, end= " ")
    print()
```

```python
n = 5

for i in range(n):
    for j in range(n):
        print(i+j+1, end=" ")
    print()
```

```python
n = 5

for i in range(n):
    for j in range(i+1):
        print(j + 1, end=" ")
    print()
```

```python
n = 5

for i in range(n):
    for j in range(i + 1):
        print(i + j + 1, end= " ")
    print()
```

```python
n = 5

for i in range(n):
    for space in range(n - i - 1):
        print(" ", end="")
    for st in range(i + 1):
        print("*", end = "")
    print()
```
