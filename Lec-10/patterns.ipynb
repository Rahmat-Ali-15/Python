```python
n = 9

for i in range(n):
    for j in range(n):
        if i == 0 or i == n-1 or j == 0 or j == n-1 or i == (n//2) or j == (n//2):
            print("1", end=" ")
        else:
            
            print("0", end = " ")
    print()
```

```python
n = 9

for i in range(n):
    for j in range(n):
        if i == 0 or i == n-1 or j == 0 or j == n-1 or i == (n//2) or j == (n//2):
            print("1", end=" ")
        else:
            
            print(" ", end = " ")
    print()
```

```python
n = 7

for i in range(n):
    for j in range(n):
        # (i = 0, j = 0) (i = 0, j = n-1) (i = n -1, j = 0) (i = n-1, j = n-1)
        if (i == 0 and j == 0) or (i == 0 and j == n-1) or (i == n-1 and j == 0) or (i == n-1 and j == n-1):
            print("+", end= " ")
        # row - 0 or row n-1 --> - 
        elif i == 0 or i == n-1:
            print("-", end=" ")
        # col = 0 or col = n-1 --> |
        elif j == 0 or j == n-1:
            print("|", end=" ")
        else:
            print(" ", end=" ")
    print()
```
