## for loops

### range(start, stop, stride/step)

```python
for item in range(1, 11, 1):
    print(item, end=" ")
```

### Default values of range() 

```python
# If we not define the value of start & step then it will take its default value which is start = 0, step = 1
for i in range(10):
    print(i, end=" ")
```

```python
# Here the default value of step = 1
for i in range(1, 11):
    print(i, end = " ") 
```

```python
for i in range(10):
    print(i + 1, end=" ")
```

```python
# this will not print anything
for i in range(10, 1, 1):
    print(i, end=" ")
```

```python
for i in range(10,0, -1):
    print(i, end=" ")
```

```python
# Print n numbers
a = int(input("Enter the number : "))
for i in range(1, a,1):
    print(i, end=" ")
```

```python
for i in range(10, 0, -2):
    print(i, end = " ") 
```

```python
# Prime Number Detector

n = int(input())

if n < 2:
    print(f"{n} is not a Prime Number")
else:
    for i in range(2, n):
        if n % i == 0:
            print(f"{n} is not a Prime Number")
            break
    else:
        print(f"{n} is a Prime Number")
```

#### Nested Loopig

```python
for outer_loop in range(1, 6):
    print(f"{outer_loop}", end=" -> ")
    for inner_loop in range(11, 14):
        print(f"{inner_loop}", end=" ")
    print()
```

```python
for i in range(5):
    for j in range(5):
        print(f"({i + 1}, {j + 1})", end="\t")
    print()
```

```python
for i in range(5):
    for j in range(5):
        print(f"(Row = {i+1}, Col = {j+1})", end="  ")
    print()
```

```python
n = int(input("Entet the number"))

for i in range(n):
    for j in range(i+1):
        print("*", end = "")
    print()
```
