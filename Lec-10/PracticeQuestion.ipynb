## 1. Print numbers in a square

```python
n = int(input())

for i in range(n):
    for j in range(n):
        print(i+1, end = " ")
    print()
```

## 2. Right angle triangle (stars)

```python
n = int(input())

for i in range(n):
    for j in range(i+1):
        print("*", end = " ")
    print()
```

## 3. Right angle triangle (numbers)

```python
n = int(input())

for i in range(n):
    for j in range(i+1):
        print(j +1, end=" ")
    print()
```

## 4. Same number triangle

```python
n = int(input())

for i in range(n):
    for j in range(i +1):
        print(i+1, end = " ")
    print()
```

## 5. Alphabet triangle

```python
n = int(input())

for i in range(n):
    for j in range(i+1):
        print(chr(65 + j), end=" ")
    print()
```

## 7. Right aligned star triangle

```python
n = int(input())

for i in range(n):
    for sp in range(n - i - 1):
        print(" ", end= " ")
    for star in range(i + 1):
        print("*", end= " ")
    print()
```
