#### Indexing

```python
li = [10, 8, 2, 4, 1, 5, 20]

print(li)
```

```python
# First Element

print(f"First Element : {li[0]}") # Prefered
print(f"First Element : {li[-7]}")
print(f"First Element : {li[-len(li)]}") # Prefered
```

```python
# Last Element

print(f"Last Element : {li[-1]}") # Prefered
print(f"Last Element : {li[6]}")
print(f"Last Element : {li[len(li) - 1]}")
```

#### Slicing

```python
print(f"Original List : {li}\n\nForward Indexing")

print(f"First 3 Elements : {li[0:3:1]}")
print(f"Last 3 Elements : {li[4:7:1]}")
print(f"Every Element except First and Last : {li[1:6:1]}")
```

```python
print(f"Original List : {li}\n\nNegative Indexing")

print(f"Frist 3 Elements : {li[-7:-4:1]}")
# print(f"Last 3 Elements : {li[]}")
# print(f"Every Element Except First and Last : {li[1:6:1]}") 
```

### Deletion

### Index Based

#### pop() => pop() is used to remove element at the given index and its default value is last index of the list

```python
li = [8,2,4,1,5]

print(f"Original List : {li}")

# Removing the last Element :

removed_number = li.pop()

print(f"Updated List : {li}")

# Removing on the basis of index (oth index)

removed_data = li.pop(0)

print(f"Updated List : {li}")
```

#### remove() => remove() is used to remove a specific element form a list. It removes the first occurrence of that value. 

```python
# Deletion - Element

li = [8,2,4,1,5,2]
print(li)

li.remove(2)
print(li)

# li.remove(1000)
li.remove(5)
print(li)
```

## Loops in List

### Index based Looping

```python
li = [8,2,1,4,5]
```

```python
for i in range(0,5,1):
    print(i, li[i])
```

```python
for i in range(len(li)):
    print(f"Index : {i}, Data : {li[i]}")
```

```python
for i in range(-1, -len(li) - 1, -1):
    print(f"Index : {i}, Data : {li[i]}")
```

### Item based

```python
for item in li:
    print(item)
```

### Enumerate based

```python
for index, item in enumerate(li):
    print(f"Index : {index}, Data : {item}")
```
