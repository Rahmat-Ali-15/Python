```python
# Print a factorial of n numbers

start = int(input("Entet the number"))
ref = start
stop = 1
fact = 1

while start>= stop:
    fact *= start
    start -= 1
print(f"{ref}! = {fact}")
```

```python
# Prime Number

n= int(input("Enter the number to check : "))

if n < 2:
    print(f"{n} is not a prime number")
else:
    start = 2

    while start < n:
        if n % start == 0:
            print(f"{n} is not a prime number")
            break
        else:
            start += 1
    else:
        print(f"{n} is a prime number")
```
